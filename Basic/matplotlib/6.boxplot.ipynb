```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset=load_dataset('lukebarousse/data_jobs')
```

```python
df=dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x:ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_da=df[(df['job_title_short']=='Data Analyst')&(df['job_country']=='United States')].copy()
df_da
```

```python
df_da=df_da.dropna(subset=['salary_year_avg'])
```

```python
df_da
```

```python
df_da['salary_year_avg'].sample(10)
```

```python
plt.boxplot(df_da['salary_year_avg'],vert=False)
plt.show()
```

```python
df_us=df[(df['job_country']=='United States')]
```

```python
df_us.dropna(subset='salary_year_avg')
```

```python
job_title=['Data Analyst','Data Engineer','Data Scientist']
```

```python
df_us=df[df['job_title_short'].isin(job_title)]
```

```python
df_us
```

```python
df_us=df_us.dropna(subset='salary_year_avg')
```

```python
job_titles=[df_us[df_us['job_title_short']==job_titl]['salary_year_avg'] for job_titl in job_title]
job_titles
```

```python
plt.boxplot(job_titles,labels=job_title,vert=False)
ax=plt.gca()
plt.xlabel('Yearly salary')
plt.title('Distribution of salary on different jobs')
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_:f'{int(x/1000)}K'))
plt.show()
```
