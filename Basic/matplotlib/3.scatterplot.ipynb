```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset=load_dataset('lukebarousse/data_jobs')
```

```python
df=dataset['train'].to_pandas()
```

```python
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_da=df[df['job_title_short']=='Data Analyst']
df_da
```

```python
df_exploded=df_da.explode('job_skills')
df_stat=df_exploded.groupby('job_skills').agg(
    skill_count=('job_skills','count'),
    median_salary=('salary_year_avg','median')
)
df_stat=df_stat.sort_values(by='skill_count',ascending=False).head(10)
```

```python
df_stat.plot(kind='scatter',x='skill_count',y='median_salary',title='Salary vs count of skill for top jobs')
for i, skill in enumerate(df_stat.index):
    plt.text(df_stat['skill_count'].iloc[i],df_stat['median_salary'].iloc[i],skill)



plt.tight_layout()    
```
