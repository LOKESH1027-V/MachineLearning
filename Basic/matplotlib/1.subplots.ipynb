```python
import pandas as pd
import matplotlib.pyplot as plt
import ast
from datasets import load_dataset

dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()

df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
```

```python
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

```python
fig, ax=plt.subplots(1,2)
```

```python
df_skills=df.copy()
```

```python
df_skills=df_skills.explode('job_skills')
```

```python
skill_count=df_skills.groupby(['job_skills','job_title_short']).size()
```

```python
df_skill_count=skill_count.reset_index(name='skill_count')
```

```python
df_skill_count.sort_values(by='skill_count',ascending=False,inplace=True)
```

```python
df_skill_count
```

```python
job_titile=['Data Analyst','Data Engineer','Data Scientist']
```

```python
fig, ax=plt.subplots(3,1)

for i, job  in enumerate(job_titile):
    df_plot=df_skill_count[df_skill_count['job_title_short']==job].head(5)
    df_plot.plot(kind='barh',x='job_skills',y='skill_count',ax=ax[i],title=job)
    ax[i].invert_yaxis()
    ax[i].set_xlim(0,120000)
    ax[i].legend().set_visible(False)
fig.suptitle(f'Skill counting for top jobs')
fig.tight_layout()
```
