```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load data
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
```

```python
df.job_posted_date=pd.to_datetime(df.job_posted_date)
```

```python
df_us=df[df['job_country']=='United States'].copy()
```

```python
df_us
```

```python
df
```

```python
df_us
```

```python
df_us['job_posted_month']=df_us['job_posted_date'].dt.strftime('%B')
```

```python
df_us_pivot=df_us.pivot_table(index='job_posted_month', columns='job_title_short',aggfunc='size')
df_us_pivot
```

```python
df_us_pivot['month_no']=pd.to_datetime(df_us_pivot.index, format='%B').month
```

```python
df_us_pivot
```

```python
df_us_pivot.sort_values('month_no' , inplace=True)
```

```python
df_us_pivot.drop(columns='month_no')
```

```python
df_us_pivot.plot(kind="line")
```

```python
top3=df_us['job_title_short'].value_counts().head(3)
top3=top3.index
```

```python
df_us_pivot[top3].plot(kind='line')
plt.title('Monthly job posted for top 3 roles')
plt.xlabel('2023')
plt.ylabel('Job count')
plt.show()
```

merge two data sets


```python
df_us_software_pivot=pd.read_csv("http://lukeb.co/software_csv")
df_us_software_pivot.set_index('job_posted_month')
```

```python
df_us_merged=df_us_pivot.merge(df_us_software_pivot, how='inner', on='job_posted_month')
df_us_merged=df_us_merged.set_index('job_posted_month')
```

```python
top5=df_us_merged.sum().sort_values(ascending=False)
top5=top5.head(5).index
```

```python
df_us_merged[top5].plot(kind='line')
plt.title("Top tech jobs in us")
plt.xlabel('2023')
plt.ylabel('Job count')
plt.ylim(0,20000)
plt.show()
```
