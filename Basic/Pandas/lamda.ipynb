```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load data
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
df['jon_posted_date']=pd.to_datetime(df['job_posted_date'])
```

```python
df[pd.notna(df['salary_year_avg'])]['salary_year_avg']
```

```python
def projected_salary(s):
    return s*0.3
df_salary=df[pd.notna(df['salary_year_avg'])].copy() 
```

```python
df_salary['salary_inflanted']=df_salary['salary_year_avg'].apply(projected_salary)
```

```python
df_salary[['salary_year_avg','salary_inflanted']]
```

```python
df_salary['s_inflanted']=df_salary['salary_year_avg'].apply(lambda salary: salary*0.3)
```

```python
df_salary['s_inflanted']
```

```python
import ast

job_skill=df[df['job_skills'].notna()]
job_skill['job_skills']=job_skill['job_skills'].apply(ast.literal_eval)
```

```python
job_skill['job_skills']
```

```python
def promoted(row):
    if  "Senior" in row:
        return 1.03*row['salary_year_avg']
    else:
        return 0.3*row['salary_year_avg']

job_skill['promoted_salary']=job_skill.apply(promoted,axis=1)
job_skill
```
