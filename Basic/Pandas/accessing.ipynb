```python
import pandas as pd
from datasets import load_dataset
d=load_dataset('lukebarousse/data_jobs')
```

```python
df=d['train'].to_pandas()
```

```python
df.info()
```

```python
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
```

```python
df.groupby('job_country')['salary_year_avg'].sum()
```

```python
df['month']=df['job_posted_date'].dt.month
```

```python
monly_count=df['month'].value_counts()
```

```python
df
```

```python
monly_count=monly_count.sort_index()
```

```python
import matplotlib.pyplot as plt
plt.plot(monly_count.index,monly_count)
```

```python
job_title_count=df.job_title_short.value_counts()
job_title_count
```

```python
plt.bar(job_title_count.index,job_title_count)
plt.show()
```

```python
job_title_count=job_title_count.sort_values(ascending=True)
plt.barh(job_title_count.index,job_title_count)
```

```python
plt.bar(job_title_count.index,job_title_count)
plt.title("job count based on title")
plt.ylabel("job count")
plt.xticks(rotation=90)
```
